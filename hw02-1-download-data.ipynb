```python
import pandas as pd
import boto3
import io
import os
```

```python
s3 = boto3.resource('s3')
s3.create_bucket(Bucket='dsan6000-em1569')
```

```python
source_bucket = s3.Bucket(name='dsan6000-wikipedia')
```

```python
parquet_files = list(source_bucket.objects.filter(Prefix='hourly_parquet'))

for summary in parquet_files:
    obj = summary.Object()

    if obj.key == "hourly_parquet/":
        continue

    file_name = os.path.basename(obj.key)
    local_folder = os.path.join("data/", file_name)
    
    source_bucket.download_file(obj.key, local_folder)
```

```python
dest_bucket = 'dsan6000-em1569'

for file in os.listdir('data/'):
    if file.endswith('.parquet'):
        s3.Bucket(dest_bucket).upload_file(f"data/{file}", f"wikipedia-hourly/{file}")
```
